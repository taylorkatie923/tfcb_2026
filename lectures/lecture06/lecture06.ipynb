### f-strings

Python gives us a SUPER convenient way of creating nicely formatted strings from other objects called `f-strings`. You don't need to memorize all the formatting options, but we'll use `f-strings` later in this notebook.

```python
# f-strings start with an f before the quotes, and contain {} expressions that are replaced with the value of
#  whatever python expression is inside them. The {} expressions can contain a ':' that is followed by cryptic instructions for
#  how python should format the thing before the ':'

name = 'Phil'
age = 29

message = f'My name is {name}, and I am {age} years old.' # yeah right
message
```

```python
# f-strings are useful for formatting numeric values nicely, among many other things
# here the :5d tells python to make the length 5, by padding with whitespace if necessary
# and :9.2e says use scientific notation (e) with a total length of (at least) 9 and 2 digits after the 
# decimal point
#
p_value = 0.000003545
num_trials = 154
print(f'number of trials: {num_trials:5d} P-value: {p_value:9.2e}')

p_value = 0.000243
num_trials = 12
print(f'number of trials: {num_trials:5d} P-value: {p_value:9.2e}')

p_value = 1.7456e-20
num_trials = 12043
print(f'number of trials: {num_trials:5d} P-value: {p_value:9.2e}')
```

## Danger

Assignment (e.g., setting `x=y`) never makes a new copy of the object that `y` points to; it just makes `x` another name for the same object. To make a copy of a list, use slicing (`x=y[:]`), the `copy` method (`x=y.copy()`), or explicit creation of a new object with the object typename (e.g. `x=list(y)`). (Dictionaries can't be sliced, so use `x=y.copy()` or `x=dict(y)` for those.)

These all make *shallow* copies: the new list is a new object, but the things *inside* it are not copied. So if a list contains other lists, the inner lists are still shared (see the second example below). `copy.deepcopy` from the `copy` module makes a fully independent copy.

```python
# create a new object of type list with the contents [1,2,3,4,5] and assign the name 'a' to point to it
a = [1,2,3,4,5]
print('a=',a)
b = a # create a new name 'b' and point it at the same object that a points to
b[0] = 1000
print('b=',b)
print('a=',a) # a was changed, too
```

```python
c = a[:] # slicing a list makes a copy
c[0] = 0
print('c=',c)
print('a=',a) # a still the same as before we modified c

d = list(a)
d[0] = -1
print('d=',d)
print('a=',a) # a still the same as before we modified d
```

```python
# shallow vs deep copies: the inner lists are shared by a shallow copy
# 'import' loads a MODULE, a bundle of extra python code (more on modules later in this notebook).
# Afterwards we can use the functions inside it with a '.', like copy.deepcopy()
import copy

nested = [[1,2], [3,4]]
shallow = list(nested) # new outer list, but it holds the SAME inner list objects
deep = copy.deepcopy(nested) # copies everything, all the way down

nested[0][0] = 1000
print('nested= ', nested)
print('shallow=', shallow) # changed too!
print('deep=   ', deep)    # unchanged
```

## OPTIONAL DIGRESSION: object introspection

In programming, introspection refers to the ability to investigate the type and/or other properties of objects at runtime. Remember that just about everything in python is an object! Two useful functions for introspection are `dir` and `id`.

`dir` shows all the methods (functions attached to the object) and attributes (accessible internal data) of an object.

`id` returns a unique identifier for the given object (for example, the location in memory where it is stored). No two objects that exist at the same time will have the same `id` (but an `id` can be reused after an object is deleted). `id` can be useful for figuring out whether two names point to the same underlying object. The `is` keyword does exactly this check for us: `a is b` is `True` if `a` and `b` point to the same object.

(One surprise: for efficiency, python sometimes re-uses a single object for small integers and short strings, so after `a=4` and `b=4`, `a is b` may well be `True`. Use `==` to compare values; use `is` only when you really want to know whether two names point to the very same object, e.g. `x is None`.)


```python
# create list
my_list = [1,2]

# show attributes and methods of my_list
print(dir(my_list))
```

```python
a = [4,5,6]
b = a # did we make a copy or not??
b.append(3)

print('id(a)=', id(a))
print('id(b)=', id(b))

# here we store the result of a comparison (checking whether two things are equal) in a boolean variable (ie, object)
the_same = id(a) == id(b)
print('a and b point to the same object:', the_same)

# the 'is' keyword is the idiomatic way to do this check
print('a is b:', a is b)

c = a[:] # an actual copy
print('c == a:', c == a) # same contents...
print('c is a:', c is a) # ...but a different object
```

# flow of control: `for` loops
Python `for` loops let you repeat a block of code while changing the value of a LOOPING VARIABLE. The code to be repeated is INDENTED. This is a really important and distinctive thing about python: the formatting of the code, specifically the level of indenting, actually affects the syntax. Some people hate this. Some people love it. I like it and think it makes code more readable. But it takes some getting used to if you are coming from other languages like C/C++, java, etc. 

```python
# a simple for-loop over the elements of the list

pets = ['dog', 'cat', 'goldfish']

# Below, 'pet' is the looping variable (name)
# it is equal to each of the elements of the list, in turn. First 'dog', then 'cat', then 'goldfish'
# In other words, the first time through the loop, pet is a name that points to the first object in the list
# the second time through the loop, pet points to the second object in the list. And so on.
#
for pet in pets:
    print('A', pet, 'is a great pet, if you happen to like pets.')
print('Those are the only pets I know.') # this only gets printed at the end; try indenting it and re-running...
```

## `for` is an incredibly versatile statement: you can loop over almost any kind of object that has multiple items or components

```python
# we can loop over the keys of a dictionary
base_partner = {'A':'T', 'T':'A', 'C':'G', 'G':'C'}

for a in base_partner:
    print(a, 'pairs with', base_partner[a])
```

We can also loop over the key,value pairs of a dictionary directly, using the `.items()` method:

```python
base_partner = {'A':'T', 'T':'A', 'C':'G', 'G':'C'}

for a, a_partner in base_partner.items():
    print(a, 'pairs with', a_partner)
    
```

```python
# we can loop over the elements of a list
words = 'the quick brown fox jumps over the lazy dog'.split() # create a list by splitting a string

counter = 0
for word in words:
    print('The', counter, 'word is', word)
    counter += 1

# we can loop over a range of numbers and use them as indices
print('\nAgain,') # using \n to add some space
for i in range(len(words)):
    print('The', i, 'word is', words[i])

# enumerate is a wonderful built-in function that gives us BOTH the index AND the item.
# This is the most 'pythonic' way to do this, and it is less error-prone than the two above
print('\nAnd again,')
for i, word in enumerate(words):
    print(f'The {i} word is {word}')
```

```python
# zip lets us loop over two (or more) lists in parallel
genes = ['TP53', 'BRCA1', 'MYC']
chromosomes = ['17', '17', '8']

for gene, chrom in zip(genes, chromosomes):
    print(f'{gene} is on chromosome {chrom}')
```

# `while` loops

`while` loops repeat a block of indented code as long as a boolean expression evaluates to `True`

```python
# SUPER-HANDY TIP:
# you can assign two (or more) variables at the same time. This is called 'unpacking' in python
# (more on unpacking in the tuples section below).
a, b = 0, 1

# generate the fibonacci sequence
while a<1000:
    print('a=',a)
    # Here unpacking saves us from having to use a temporary variable
    a, b = b, a+b
```

Functions can return results to the caller, using the `return` statement

```python
base_partner = {'A':'T', 'T':'A', 'C':'G', 'G':'C'}
num_hbonds = {'A':2, 'T':2, 'C':3, 'G':3}


def reverse_complement(seq):
    "returns the reverse complement of a nucleic acid sequence"
    bwd = ''
    for base in seq:
        bwd = base_partner[base] + bwd
    return bwd

def total_hbonds(seq):
    'returns the total number of watson-crick hbonds between seq and its reverse complement'
    total=0
    for a in seq:
        total += num_hbonds[a]
    return total

fwd = 'ACGGTAATGATCCTCAG'
rev = reverse_complement(fwd)

print('fwd=', fwd, 'rev=', rev, 'num_hbonds=', total_hbonds(fwd))

# an assert statement checks to see if something is true, and will interrupt the program if it's not.
assert reverse_complement(reverse_complement(fwd)) == fwd
assert total_hbonds(fwd) == total_hbonds(rev)
```

Functions can have OPTIONAL ARGUMENTS whose DEFAULT VALUES are pre-specified in the function definition.

```python
def reverse_complement(seq, unk_partner='N'):
    """Returns the reverse complement of a nucleic acid sequence
    
    Uses unk_partner as the partner of unrecognized letters
    """

    rseq = ''
    for a in seq:
        if a in base_partner:
            rseq = base_partner[a] + rseq
        else:
            rseq = unk_partner + rseq
    return rseq

fwd = 'ACTGTAGCxGAcTNCGAC'
print(reverse_complement(fwd))
print(reverse_complement(fwd, unk_partner='-'))
```

Functions can call other functions, even themselves:

```python
def factorial(n):
    """Calculate the factorial of a number recursively. Bad things will happen 
    if the number is negative or not an integer """
    if n==0:
        return 1
    else:
        return n * factorial(n-1) # this is called "recursion"
    
# the help function prints the docstring
help(factorial)

for i in range(10):
    print(i,'factorial =',factorial(i))
```

# Using functions from the standard library

Python comes with an extensive *standard library* of **modules** which contain all sorts of useful special purpose functions and objects. A few that are especially handy for computational biology:

* `math`: mathematical functions and constants (`math.sqrt`, `math.log`, `math.pi`, ...)
* `random`: random numbers, shuffling, and random sampling
* `statistics`: simple summary statistics (`mean`, `median`, `stdev`, ...)
* `collections`: extra data structures, e.g. `Counter` for counting things (like k-mers) and `defaultdict`
* `itertools`: tools for looping, e.g. all the `combinations` or `product`s of a set of items
* `re`: regular expressions, for searching for patterns in strings (like sequence motifs)
* `pathlib` and `os`: working with files, directories, and paths (e.g., `pathlib.Path('data.txt').exists()`)
* `glob`: finding files with wildcards (like using `*` on the command line)
* `csv`, `json`, `gzip`: reading and writing tabular, JSON, and compressed (`.gz`) files
* `sys`: information about the running python, e.g. `sys.argv`, the command-line arguments to a script
* `subprocess`: running other programs (like command-line bioinformatics tools) from python
* `argparse`: handling command-line options when you turn your code into a script
* `copy`: copying objects (we used `copy.deepcopy` above)
* `timeit` and `pdb`: timing code snippets, and the python debugger (also available via the built-in function `breakpoint()`)

[The full list is here](https://docs.python.org/3/library/). Many of the most important tools for data analysis (numpy, scipy, pandas, matplotlib, scikit-learn, ...) are *not* part of the standard library: they are separate *packages* that have to be installed (e.g. with `conda` or `pip`), but once they're installed we `import` them in exactly the same way.

```python
# Here we are "importing" a module called math which contains some math-y functions like sqrt and floor
import math


def is_prime(num):
    """Figure out whether num is prime"""
    assert isinstance(num, int) # isinstance is the standard way to check an object's type
    assert num >= 1
    if num == 1:
        return False # 1 is not prime, by definition
    
    max_factor = math.floor(math.sqrt(num))
    #max_factor = num-1
    
    for i in range(2,max_factor+1):
        if num%i == 0:
            return False
    return True

l=[]
for i in range(1,100):
    if is_prime(i):
        print(i)
        l.append(i)

print(len(l))

        
# look at the distribution of primes less than 1000

prime_counts_per_100 = [0]*10 # we need 10 bins for this

for i in range(1,1000):
    if is_prime(i):
        counts_bin = i//100
        prime_counts_per_100[counts_bin] += 1
        
print('prime_counts_per_100:', prime_counts_per_100)
```

```python
def is_prime_v2(num, smaller_primes):
    """Figure out if num is prime based on the set of smaller primes"""
    max_factor = math.floor(math.sqrt(num))
    for p in smaller_primes:
        if p>max_factor: break
        if num%p == 0:
            return False
    return True


# test the prime number theorem which says that the ratio between the number of primes
# less than N and N/log(N) approaches 1 as N gets large (math.log is the natural log)
#

primes = []

for num in range(2,10000):
    if is_prime_v2(num, primes):
        primes.append(num)
    if num%100==0:
        actual = len(primes)
        estimate = num / math.log(num)
        ratio = actual / estimate
        print(f'N= {num:5d} actual= {actual:4d} estimate= {estimate:7.1f} ratio= {ratio:.3f}')


[3,5,7]
31
            
```

# list comprehension
Python gives us a super-compact way of creating lists called a *list comprehension*. It looks kind of like a for-loop inside square brackets.

The general syntax is `[<value> for <var> in <list> if <condition>]` where the `if <condition>` part is optional.


```python
even_numbers = [2*i for i in range(50)]
print(even_numbers)
```

```python
even_numbers = [i for i in range(100) if i%2==0]
print(even_numbers)
```

```python
base_partner = {'A':'T', 'T':'A', 'C':'G', 'G':'C'}

fwd = 'ACATGTCGATCGTAGCTGCA'

# recall the string .join method from above, which joins together a list of strings into a single string
rev = ''.join([base_partner[x] for x in reversed(fwd)]) # reversed lets us loop over the elements of a list/string/etc in reverse order
rev
```

# `tuples`

Dictionary keys must be *hashable*. For the built-in types this basically means *immutable*: objects (like integers, floats, strings) that can't be changed after they are created. So lists are no good. But, there's something called a `tuple` which looks a lot like a list and can be used as a dictionary key (as long as everything inside it is immutable too), and in lots of other places where multiple items are passed around.

```python
# create a tuple
t = (4,5,'a')

# create an empty dictionary
d = {}

# set the value 4 for the key t in the dictionary d
d[t] = 4

# print stuff
print('t=', t)
print('d=', d)
```

`tuples` can be **unpacked** by assigning to them a sequence of variables matching their length: 

```python
nums = ( 4, 11, 17 )

a,b,c = nums

print('nums=',nums)
print('a=',a)
print('b=',b)
print('c=',c)
```

# Notebook pitfalls

Notebooks are great for exploring, but they have a catch: the kernel remembers everything you've run, **in the order you ran it**, which isn't necessarily the order of the cells on the page. That can make a notebook look correct when it isn't:

* **Out-of-order execution.** If you go back and edit or re-run an earlier cell, later cells may now be working with different values than you'd expect from reading top to bottom. The number in `[ ]` next to each code cell shows the order the cells were actually run.
* **Deleted code lives on.** If you delete or rename a variable or function, the old one is still in the kernel's memory. Code that seems to work may depend on something that's no longer anywhere in the notebook, and it will fail for the next person who runs it (or for you, tomorrow). The `Variables` button in the VS Code notebook toolbar shows what's currently in memory.
* **Re-running isn't always harmless.** A cell that *changes* something (like `append`ing to a list) changes it again every time you run it. Try the two cells below.

**The fix:** before you trust a result, or turn in a homework, restart the kernel and run everything from the top (VS Code: `Restart`, then `Run All`, in the notebook toolbar; browser Jupyter: `Kernel` → `Restart Kernel and Run All Cells`). If the notebook runs cleanly from top to bottom and gives the same answers, it's reproducible.

```python
# Run this cell once...
pets = ['dog', 'cat']
```

```python
# ...then run this cell three times. Then run the cell above again, followed by this one.
pets.append('hamster')
print(pets)
```

# A note on AI coding assistants

AI assistants (e.g., GitHub Copilot, ChatGPT, Claude, Gemini) can write, explain, and debug python code, and they can be built right into VS Code. Many working scientists use them every day; lecture 14 covers them in more depth. A few suggestions for using them while you're learning:

* **Use them as a tutor.** Pasting in an error message or a confusing bit of code and asking "what does this do, and why?" is a great way to learn.
* **You are responsible for the code.** AI-generated code can run without errors and still be wrong in ways that matter for biology: off-by-one errors in sequence coordinates, the wrong strand or reading frame, silently skipping bad input. Read the code, and test it on small examples where you know the right answer (like the `assert` statements in our `reverse_complement` example).
* **Learn the fundamentals anyway.** You need them to tell whether the code an assistant gives you is correct. Try the practice problems in this notebook on your own first.
* **Be careful with data.** Don't paste patient data, unpublished results, or passwords into an online tool unless your institution has approved it for that use.
* **Cite it.** The course homework policy asks you to cite any source that helped you in comments in your code; AI assistants count too.
